```python
import numpy as np

arr = np.random.randint(1, 21, size=(5, 5))

print("Original Array:")
print(arr)

arr[:, 2] = 1

print("Modified Array:")
print(arr)
```

```python
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

np.fill_diagonal(arr, 0)

print("Modified Array:")
print(arr)
```

```python
import numpy as np

arr = np.random.randint(1, 21, size=(5, 5))

print("Original Array:")
print(arr)

border = np.concatenate((
    arr[0, :],
    arr[-1, :],
    arr[1:-1, 0],
    arr[1:-1, -1]
))

print("Border Elements:")
print(border)
```

```python
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print("Array:")
print(arr)

print("Row-wise Sum:")
print(np.sum(arr, axis=1))

print("Column-wise Sum:")
print(np.sum(arr, axis=0))
```

```python
import numpy as np

arr = np.arange(1, 10).reshape(3, 3)

print("Original Array:")
print(arr)

a = arr.reshape(1, 9)
print("Shape (1, 9):")
print(a)

b = arr.reshape(9, 1)
print("Shape (9, 1):")
print(b)
```

```python
import numpy as np

data = np.array([
    ("Yash", 22, 65.5),
    ("Sunil", 20, 70.2),
    ("Rahul", 21, 68.4),
    ("Amit", 19, 60.5)
], dtype=[
    ("name", "U20"),
    ("age", "i4"),
    ("weight", "f4")
])

print("Original Array:")
print(data)

sorted_data = np.sort(data, order="age")

print("Sorted by Age:")
print(sorted_data)
```
